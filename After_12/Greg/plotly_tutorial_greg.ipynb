```python
import pandas as pd
url = "https://github.com/alex-degarate/DAnalytics/raw/refs/heads/main/After_13/Greg/"

# el archivo original no esta delimitado por comas
df = pd.read_csv( url + "AnnualTicketSales.csv")
df.head()
```

```python
# el csv presenta varios problemas:
#  los campos no estan separados por comas
#  el nombre de los campos contienen espacios
# varios campos contienen ',' y '$'

#df.drop(df.columns[-1], axis=1, inplace=True)
#df.drop('TOTAL_INFLATION_ADJUSTED_BOX_OFFICE', axis=1, inplace=True)

def fix_columns(x):
  x = str(x)
  x = x.replace(',', '')
  x = x.replace('$', '')

  return int(float(x))

for column in df.columns:
  df[column] = df[column].apply(fix_columns)

df.head()
```

```python
import plotly.express as px

px.pie(df,
       names='AVERAGE_TICKET_PRICE')
```

```python
px.histogram(df, x='TICKETS_SOLD', nbins=2, title='Histogram of Tickets Sold')
```

```python
fig = px.box(df,
       y='TOTAL_BOX_OFFICE',
       title='Box & Whisker Plot of Total Box Office',
       hover_data=['YEAR'],
       )
fig.update_traces(fillcolor='#ffffcc', line=dict(color='navy', width=2))

fig.show()
```

```python
fig = px.scatter(df, x='YEAR',
           y='TICKETS_SOLD',
           title='Movie Ticket Sales vs Time')

fig.update_traces(marker=dict(size=16,
                              line=dict(width=2,
                                        color='DarkSlateGrey')),
                  selector=dict(mode='markers'))

fig.show()
```

```python
px.bar(df, x='YEAR', y='TOTAL_BOX_OFFICE', color='AVERAGE_TICKET_PRICE')
```

```python
fig = px.scatter(df, x='YEAR',
           y='TICKETS_SOLD',
           size='AVERAGE_TICKET_PRICE',
           title='Movie Ticket Sales vs Time',
           labels={'TICKETS_SOLD':'Tickets Sold',
                   'YEAR':'Year'}
           )

fig.update_traces(marker=dict(
            color='Green',
              line=dict(
              color='Blue',
            width=2
            )
        ),)

fig.update_layout(
    font_family="Calibri",
    font_color="blue",
    font_size=18,
    title_font_family="Bangers",
    title_font_size=48,
    title_font_color="Red",
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

x = np.arange(-30, 30)

fig.add_trace(go.Scatter(x=x, y=x**1, mode='markers', name='Linear'))

fig.add_trace(go.Scatter(x=x, y=x**2, mode='markers', name='Quadratic'))

fig.update_layout(title="Linear vs Quadratic Functions!")

fig.show()
```

```python
from plotly.subplots import make_subplots

fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Linear", "Square", "Cube", "Quarted"))

x = np.arange(-30, 30)

fig.add_trace(go.Scatter(x=x, y=x**1, mode='markers'),
              row=1, col=1)

fig.add_trace(go.Scatter(x=x, y=x**2, mode='markers'),
              row=1, col=2)

fig.add_trace(go.Scatter(x=x, y=x**3, mode='markers'),
              row=2, col=1)

fig.add_trace(go.Scatter(x=x, y=x**4, mode='markers'),
              row=2, col=2)

fig.update_layout(showlegend=False, title="What Nice Graphs!")


fig.show()
```
