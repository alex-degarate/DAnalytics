```python
import pandas as pd
import cufflinks as cf
from IPython.display import display,HTML
import plotly.express as px

# write cf.getThemes() to find all themes available
cf.set_config_file(sharing='public',theme='ggplot',offline=True)
cf.go_offline()
cf.getThemes()
```

# The Data

```python
#https://github.com/thepycoach/data-visualization/blob/main/datasets/population/population_total.csv

url = "https://raw.githubusercontent.com/thepycoach/data-visualization/main/datasets/population/"
df_population = pd.read_csv(url +'population_total.csv')
df_population.info()
```

```python
df_population = df_population.dropna()
df_population["year"] = df_population["year"].astype(int)
df_population["population"] = df_population["population"].astype(int)

display( df_population.head() )
```

```python
# If 'year' is already the index (from a previous run), reset it to make it a column again
if df_population.index.name == 'year':
    df_population = df_population.reset_index()

df_population.info()
```

```python
#url = "https://raw.githubusercontent.com/thepycoach/data-visualization/main/datasets/population/"
#df_population = pd.read_csv(url +'population_total.csv')

df_population.head()

df_population = df_population.pivot(index='year', columns='country',values='population')
df_population = df_population[['United States', 'India', 'China',
                               'Indonesia', 'Brazil']]

# Convert population columns to nullable integer type
for col in df_population.columns:
    df_population[col] = df_population[col].astype(pd.Int64Dtype())

df_population.info()
```

# Barplot

## Single Barplot

```python
df_population_2020_for_bar = df_population[df_population.index.isin([2020.0])].T.reset_index()
df_population_2020_for_bar.columns = ['Country', 'Population']
fig = px.bar(df_population_2020_for_bar, x='Country', y='Population',
             title='Population in 2020')
display(fig)
```

## Barplot grouped by “n” variables

```python
# filter years out
df_population_sample = df_population[df_population.index.isin([1980, 1990, 2000, 2010, 2020])]

# plotting
df_population_sample.iplot(kind='bar', xTitle='Years',
                           yTitle='Population')
```

# Boxplot

## Single Boxplot

```python
df_population['United States'].iplot(kind='box', color='green',
                                     yTitle='Population')
```

## Multiple Boxplot

```python
df_population.iplot(kind='box', xTitle='Countries',
                    yTitle='Population')
```

# Histogram

```python
df_population[['United States', 'Indonesia']].iplot(kind='hist',
                                                    xTitle='Population')
```

# Piechart

```python
# transforming data
df_population_2020_for_pie = df_population[df_population.index.isin([2020.0])].T.reset_index()
df_population_2020_for_pie.columns = ['country', '2020']

# plotting
fig = px.pie(df_population_2020_for_pie, names='country',
             values='2020', title='Population in 2020 (%)')
display(fig)
```
